```python
import pandas as pd
import pathlib, urllib.request

RAW = pathlib.Path("raw");     RAW.mkdir(exist_ok=True)
OUT = pathlib.Path("output");  OUT.mkdir(exist_ok=True)
```

```python
trip_url = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2025-01.parquet"
zone_url = "https://d37ci6vzurychx.cloudfront.net/misc/taxi_zone_lookup.csv"

trip_path = RAW / "yellow_tripdata_2025-01.parquet"
zone_path = RAW / "taxi_zone_lookup.csv"

if not trip_path.exists():
    urllib.request.urlretrieve(trip_url, trip_path)
if not zone_path.exists():
    urllib.request.urlretrieve(zone_url, zone_path)
```

```python
cols = ["tpep_pickup_datetime","tpep_dropoff_datetime","passenger_count",
        "trip_distance","PULocationID","DOLocationID","payment_type",
        "fare_amount","tip_amount","total_amount"]

trips = pd.read_parquet(trip_path, columns=cols)
zones = pd.read_csv(zone_path, keep_default_na=False)
```

```python
t = trips.copy()
```

```python
t
```

```python
t["duration_min"] = ( 
t["tpep_dropoff_datetime"] - t["tpep_pickup_datetime"] 
).dt.total_seconds().div(60) 
```

```python
valid = t.loc[ 
(t["trip_distance"] > 0) & 
(t["duration_min"] > 0) & 
(t["fare_amount"] >= 0) & 
(t["total_amount"] >= 0), 
:
].copy() 
invalid = t.loc[~t.index.isin(valid.index)].copy() 
```

```python
valid["pickup_date"] = valid["tpep_pickup_datetime"].dt.date 
```

```python
valid["pickup_hour"] = valid["tpep_pickup_datetime"].dt.hour 
```

```python
valid["pickup_weekday"] = valid["tpep_pickup_datetime"].dt.day_name() 
```

```python
valid["speed_mph"] = valid["trip_distance"].div(valid["duration_min"]/60) 
```

```python
valid["tip_pct_of_fare"] = valid["tip_amount"].div(valid["fare_amount"]).where(valid["fare_amount"].gt(0)) 
```

```python
valid
```

```python
hour_kpi = ( 
    valid.groupby("pickup_hour", as_index=False) 
         .agg(trips=("trip_distance","size"), 
              avg_distance=("trip_distance","mean"), 
              avg_duration=("duration_min","mean"), 
              total_revenue=("total_amount","sum")) 
         .sort_values("trips", ascending=False) 
) 
```
